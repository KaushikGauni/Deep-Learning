```python
# ============================================================
# TASK 1: Image Captioning (VGG16 + SimpleRNN)
# ============================================================

import os, string, pickle, gc
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import tensorflow as tf

from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Input, Dense, Embedding, SimpleRNN, Dropout, add
from nltk.translate.bleu_score import corpus_bleu

# ===================== DRIVE =====================
from google.colab import drive
drive.mount('/content/drive')

# ===================== PATHS =====================
images_path = "/content/drive/MyDrive/Images"
captions_path = "/content/drive/MyDrive/captions.txt"
names_path = "/content/drive/MyDrive/image_names.txt"
glove_path = "/content/drive/MyDrive/glove.6B.200d.txt"

# ===================== LOAD IMAGE NAMES =====================
with open(names_path) as f:
    image_names = [line.strip() for line in f if line.strip()]
valid_images = set(image_names)
print(f"Images in image_names.txt: {len(image_names)}")

# ===================== LOAD CAPTIONS =====================
mapping = {}
with open(captions_path) as f:
    for line in f:
        if '\t' not in line: continue
        img, cap = line.split('\t', 1)
        img = img.split('#')[0]
        if img in valid_images:
            mapping.setdefault(img, []).append(cap.strip())

print(f"Images with captions: {len(mapping)}")
print(f"Total captions: {sum(len(v) for v in mapping.values())}")

# ===================== CLEAN =====================
def clean(c):
    c = c.lower()
    c = c.translate(str.maketrans('', '', string.punctuation))
    return "startseq " + " ".join(c.split()) + " endseq"

for k in mapping:
    mapping[k] = [clean(c) for c in mapping[k]]

# ===================== TOKENIZER =====================
all_caps = [c for caps in mapping.values() for c in caps]
tokenizer = Tokenizer(oov_token='unk')
tokenizer.fit_on_texts(all_caps)

vocab_size = len(tokenizer.word_index) + 1
max_length = max(len(c.split()) for c in all_caps)

print("Vocab Size:", vocab_size, "Max Length:", max_length)

# ===================== FEATURES (CACHE) =====================
def extract_features(image_list):
    cache_file = "/content/features.pkl"

    if os.path.exists(cache_file):
        print("Loading cached features...")
        return pickle.load(open(cache_file, 'rb'))

    print("Extracting features...")
    model = VGG16()
    model = Model(inputs=model.inputs, outputs=model.layers[-2].output)

    features = {}
    for i, img_name in enumerate(image_list):
        path = os.path.join(images_path, img_name)
        img = load_img(path, target_size=(224,224))
        img = img_to_array(img)
        img = preprocess_input(img.reshape((1,224,224,3)))
        features[img_name] = model.predict(img, verbose=0)[0]
        if (i+1) % 500 == 0:
            print(f"  {i+1}/{len(image_list)} done")

    pickle.dump(features, open(cache_file, 'wb'))
    return features

features = extract_features(list(mapping.keys()))
mapping = {k: v for k, v in mapping.items() if k in features}
print(f"Final dataset: {len(mapping)} images")

# ===================== GLOVE =====================
embedding_dim = 200
embeddings_index = {}

with open(glove_path, encoding='utf-8') as f:
    for line in f:
        vals = line.split()
        word = vals[0]
        vec = np.asarray(vals[1:], dtype='float32')
        if len(vec) == embedding_dim:
            embeddings_index[word] = vec

embedding_matrix = np.zeros((vocab_size, embedding_dim))
for word, i in tokenizer.word_index.items():
    vec = embeddings_index.get(word)
    if vec is not None:
        embedding_matrix[i] = vec

# ===================== SPLIT =====================
keys = list(mapping.keys())
np.random.seed(42)
np.random.shuffle(keys)
split = int(0.9 * len(keys))

train_keys = keys[:split]
test_keys = keys[split:]

train_mapping = {k: mapping[k] for k in train_keys}
test_mapping = {k: mapping[k] for k in test_keys}
print(f"Train: {len(train_mapping)}, Test: {len(test_mapping)}")

# ===================== DATASET =====================
def create_dataset(mapping):
    X1, X2, y = [], [], []
    for key, caps in mapping.items():
        feature = features[key]
        for cap in caps:
            seq = tokenizer.texts_to_sequences([cap])[0]
            for i in range(1, len(seq)):
                X1.append(feature)
                X2.append(pad_sequences([seq[:i]], maxlen=max_length)[0])
                y.append(seq[i])
    return np.array(X1, dtype=np.float32), np.array(X2, dtype=np.int32), np.array(y, dtype=np.int32)

print("Building dataset...")
X1_train, X2_train, y_train = create_dataset(train_mapping)
print(f"Training samples: {len(y_train)}")

# ===================== MODEL =====================
inputs1 = Input(shape=(4096,))
fe = Dense(50, activation='relu')(Dropout(0.5)(inputs1))

inputs2 = Input(shape=(max_length,))
se = SimpleRNN(50)(Dropout(0.5)(
    Embedding(vocab_size, embedding_dim,
              weights=[embedding_matrix],
              trainable=False)(inputs2)
))

decoder = add([fe, se])
decoder = Dense(50, activation='relu')(decoder)
outputs = Dense(vocab_size, activation='softmax')(decoder)

model = Model([inputs1, inputs2], outputs)
model.compile(loss='sparse_categorical_crossentropy', optimizer='adam')

model.summary()

# ===================== TRAIN =====================
history = model.fit(
    [X1_train, X2_train], y_train,
    epochs=20,
    batch_size=64,
    verbose=1
)

# Free RAM
del X1_train, X2_train, y_train
gc.collect()

# ===================== LOSS PRINT =====================
print("\nLoss per Epoch:")
for i, loss in enumerate(history.history['loss']):
    print(f"Epoch {i+1}: {loss:.4f}")

# ===================== LOSS GRAPH =====================
plt.figure()
plt.plot(history.history['loss'])
plt.title("Loss vs Epochs")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

# ===================== GENERATE =====================
def generate_caption(feature):
    idx_to_word = {v:k for k,v in tokenizer.word_index.items()}
    text = "startseq"
    feature = feature.reshape(1,4096)

    for _ in range(max_length):
        seq = tokenizer.texts_to_sequences([text])[0]
        seq = pad_sequences([seq], maxlen=max_length)

        y_pred = model.predict([feature, seq], verbose=0)
        y_idx = np.argmax(y_pred)

        word = idx_to_word.get(y_idx)
        if word is None or word == "endseq":
            break

        text += " " + word

    return text

# ===================== BLEU =====================
def evaluate(mapping, name, max_images=None):
    actual, predicted = [], []
    items = list(mapping.items())
    if max_images:
        items = items[:max_images]
    total = len(items)

    for count, (img, caps) in enumerate(items, 1):
        y_pred = generate_caption(features[img])
        pred = y_pred.split()[1:]  # remove startseq

        refs = [c.replace('startseq ', '').replace(' endseq','').split()
                for c in caps]

        actual.append(refs)
        predicted.append(pred)

        if count % 50 == 0 or count == total:
            print(f"  {name}: {count}/{total} done...")

    print(f"\n{name} BLEU Scores:")
    print("BLEU-1:", corpus_bleu(actual, predicted, weights=(1,0,0,0)))
    print("BLEU-2:", corpus_bleu(actual, predicted, weights=(0.5,0.5,0,0)))
    print("BLEU-3:", corpus_bleu(actual, predicted, weights=(0.33,0.33,0.33,0)))
    print("BLEU-4:", corpus_bleu(actual, predicted, weights=(0.25,0.25,0.25,0.25)))

evaluate(train_mapping, "TRAIN", max_images=200)
evaluate(test_mapping, "TEST")

# ===================== SHOW EXAMPLES =====================
def show_examples(mapping, title):
    print("\n", title)
    for img in list(mapping.keys())[:5]:
        image = Image.open(os.path.join(images_path, img))
        plt.imshow(image)
        plt.axis('off')
        plt.show()

        print("Ground Truth:")
        for c in mapping[img]:
            print("-", c)

        print("Predicted:", generate_caption(features[img]))
        print("="*50)

show_examples(train_mapping, "TRAIN EXAMPLES")
show_examples(test_mapping, "TEST EXAMPLES")
```
